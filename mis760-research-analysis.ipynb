```python
import json
from pathlib import Path
import requests
import pandas as pd

DATA_DIR = Path.home() / "Downloads" / "mis760-research-data"
ATTACK_PATH = DATA_DIR / "enterprise-attack-v20.0-techniques.xlsx"
ATTACK_URL = "https://attack.mitre.org/docs/attack-excel-files/v20.0/enterprise-attack/enterprise-attack-v20.0-techniques.xlsx"
KEV_PATH = DATA_DIR / "known_exploited_vulnerabilities.json"
KEV_URL = "https://www.cisa.gov/sites/default/files/feeds/known_exploited_vulnerabilities.json"

# Download a file from the official website only if it isn't already in the folder
for path, url in [(ATTACK_PATH, ATTACK_URL), (KEV_PATH, KEV_URL)]:
    if not path.exists():
        r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=120)
        r.raise_for_status()
        path.write_bytes(r.content)
        print("Downloaded", url)

# MITRE ATT&CK Enterprise techniques (Excel file from attack.mitre.org)
attack = pd.read_excel(ATTACK_PATH, sheet_name=0)  # first sheet = techniques
print("MITRE ATT&CK:", attack.shape[0], "rows,", attack.shape[1], "columns")
display(attack.head())

# CISA Known Exploited Vulnerabilities (JSON file from cisa.gov)
with open(KEV_PATH) as f:
    kev_json = json.load(f)
kev = pd.DataFrame(kev_json["vulnerabilities"])
print("CISA KEV catalog version", kev_json["catalogVersion"], "|", kev.shape[0], "rows,", kev.shape[1], "columns")
display(kev.head())
```
